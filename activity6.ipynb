```python
import numpy as np
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt 
import warnings 
import folium
from folium.plugins import HeatMap
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('datasets\\Global_terrorist.csv', encoding='ISO-8859-1')
```

<h1 style="text-align: center; color: blue;">REGIONAL ANALYSIS</h1>


<h1>FILTERING DATA</h1>

```python
region1 = df[df['region_txt'] == 'Southeast Asia']
```

```python
region2 = region1[region1['iyear'] >= 2004]
```

```python
region3 = region2[region2['attacktype1_txt'] == 'Armed Assault']
```

```python
region4 = region3[region3['nkill'] >= 10]
```

<h1>DATA AGGREGATION</h1>

```python
region_final = region4.groupby(['latitude', 'longitude']).size().reset_index(name = 'incident_count')
region_final
```

<h1>Use of FOLIUM FOR GEOSPATIAL</h1>

<h3> Converting Pandas Data Frame into Python List</h3>

```python
region_final["normalize_data"] = region_final['incident_count'] - region_final['incident_count'].min() / \
                                region_final['incident_count'].max() - region_final['incident_count']

region_heatmap = region_final[['latitude', 'longitude', 'incident_count']].values.tolist()
```

<h3> Creating a base map </h3>

```python
region_map = folium.Map(location=[10, 105], zoom_start=4)
```

<h3> Add Heatmap Layer </h3>

```python
HeatMap(region_heatmap).add_to(region_map)
```

```python
region_map.save('regional_analysis.html')
print("data save to regional_analysis.html")
```

```python
region_map
```
