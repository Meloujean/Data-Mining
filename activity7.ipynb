<h1 style="text-align: center; color: blue;">THEFT ANALYSIS</h1>


<h1>FILTERING DATA</h1>

```python
theft1 = chicago[chicago['Primary Type'] == 'THEFT']
theft2 = theft1[theft1['Location Description'] == 'STREET']
```

<h1>DATA AGGREGATION</h1>

```python
theft3 = theft2.groupby(['Latitude', 'Longitude']).size().reset_index(name= 'incident_count')
theft3
```

<h3> Converting Pandas Data Frame into Python List</h3>

```python
theft3["normalize_data"] = theft3['incident_count'] - theft3['incident_count'].min() / \
                                theft3['incident_count'].max() - theft3['incident_count']

theft_heatmap = theft3[['Latitude', 'Longitude', 'incident_count']].values.tolist()
```

<h3> Creating a base map </h3>

```python
theft_map =  folium.Map(location = [42, -87], zoom_start = 7)
```

<h3> Add Heatmap Layer </h3>

```python
HeatMap(theft_heatmap).add_to(theft_map)
```

```python
theft_map.save('theft.html')
print('data save to theft.html')
```

```python
theft_map
```
